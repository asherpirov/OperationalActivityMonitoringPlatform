```python
north_readings = df2[df2["source_id"] == "station-north-01"]
```

```python
north_readings.head()
```

```python
north_values = pd.to_numeric(north_readings["value"], errors="coerce")
```

```python
print("north values count:", north_values.count())
print("north values min:", north_values.min())
print("north values max:", north_values.max())
print("north values mean:", north_values.mean())
```

```python
east_readings = df2[df2["source_id"] == "station-east-01"]
```

```python
east_values = pd.to_numeric(east_readings["value"], errors="coerce")
```

```python
print("east values count:", east_values.count())
print("east values min:", east_values.min())
print("east values max:", east_values.max())
print("east values mean:", east_values.mean())
```

```python
north_datetime = pd.to_datetime(north_readings["timestamp"], errors="coerce")
```

```python
north_datetime.isna().sum()
```

```python
north_readings[north_datetime.isna()]
```

```python
parsed_dates = pd.to_datetime(df2["timestamp"], errors="coerce")

parsed_values = pd.to_numeric(df2["value"], errors="coerce")
```
